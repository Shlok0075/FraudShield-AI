# 02 - Deep Exploratory Data Analysis (Week 2)

Week 2 -- deeper EDA: distributions, correlation, outliers, missing/duplicate check, time patterns.

```python
import pandas as pd, numpy as np
df = pd.read_csv('../data/raw/creditcard.csv')
df['hour_of_day'] = (df['Time'] % 86400) // 3600
df[['Amount','Time']].describe()
```

```python
df['Class'].value_counts()
```

```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
fig, ax = plt.subplots(1,2, figsize=(10,4))
df['Amount'].hist(bins=50, ax=ax[0]); ax[0].set_title('Amount distribution')
df['hour_of_day'].hist(bins=24, ax=ax[1]); ax[1].set_title('Transactions by hour')
plt.tight_layout()
plt.savefig('../../screenshots/eda/amount_and_hour_distribution.png', dpi=110)
print('saved plot')
```

```python
corr = df[[f'V{i}' for i in range(1,29)] + ['Class']].corr()['Class'].sort_values()
print('Most negatively correlated with Class:'); print(corr.head(5))
print('Most positively correlated with Class:'); print(corr.tail(5))
```

```python
print('Missing values per column:')
print(df.isnull().sum().sum())
print('Duplicate rows:', df.duplicated().sum())
```

```python
q1, q3 = df['Amount'].quantile(0.25), df['Amount'].quantile(0.75)
iqr = q3 - q1
outliers = df[(df['Amount'] > q3 + 1.5*iqr)]
print(f'IQR outliers in Amount: {len(outliers)} rows ({len(outliers)/len(df)*100:.2f}%)')
```
